Hour 1 — Exceptions

```python
numbers = [10, 20, 30]
```

```python
numbers[10]
```

```python
try:
    value = numbers[10]

except IndexError:
    print("That index does not exist")
```

```python
try:
    value = numbers[10]

except IndexError as e:
    print(f"Error occurred: {e}")
```

```python
data = {
    "product": "Laptop",
    "price": 1200
}
```

```python
try:
    value = data["quantity"]

except KeyError as e:
    print(f"Missing key: {e}")
```

else


```python
try:
    value = data["product"]

except KeyError as e:
    print(f"Missing key: {e}")

else:
    print(f"Found value: {value}")
```

finally

```python
try:
    value = data["product"]

except KeyError as e:
    print(f"Missing key: {e}")

else:
    print(f"Found value: {value}")

finally:
    print("Operation finished")
```
