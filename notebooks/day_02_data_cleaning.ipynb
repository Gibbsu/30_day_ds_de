Hour 1 — Final Cleaning Challenge

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 2, 3, 4, 5],
    "name": [" alex ", "MARIA", "MARIA", "James ", "sarah", "David"],
    "department": ["sales", "FINANCE", "FINANCE", "Sales", "hr", "HR"],
    "salary": [70000, 85000, 85000, None, 65000, 72000],
    "hire_date": [
        "2024-01-10",
        "2023-06-15",
        "2023-06-15",
        "2025-02-01",
        "2024-08-20",
        "2025-05-10"
    ]
})

display(employees)
```

Your job is to clean it so that:  

- Duplicate rows are removed.  
- name has no leading/trailing spaces and uses title case.  
- department is uppercase.  
- hire_date becomes a proper datetime.  
- Rows with a missing salary are removed.  

```python
employees.shape
```

```python
employees.info()
```

```python
employees.dtypes
```

```python
#remove duplicates
employees = employees.drop_duplicates()
```

```python
employees.shape
display(employees)
```

```python
#name has no leading/trailing spaces and uses title case.
employees["name"] = employees["name"].str.strip().str.title()
display(employees)
```

```python
#department is uppercase.#
employees["department"] = employees["department"].str.strip().str.upper()
display(employees)
```

```python
#hire_date becomes a proper datetime.
employees['hire_date'] = pd.to_datetime(employees['hire_date'])
employees.dtypes
```

```python
#Rows with a missing salary are removed.
employees = employees.dropna(subset=["salary"])
display(employees)
```

# Hour 2: Pandas transformations — calculated columns

```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "customer": ["Alex", "Maria", "James", "Sarah", "Alex", "Maria", "James", "Alex"],
    "state": ["CA", "CA", "TX", "NY", "CA", "CA", "TX", "CA"],
    "product": ["Monitor", "Laptop", "Keyboard", "Monitor", "Laptop", "Keyboard", "Monitor", "Keyboard"],
    "quantity": [2, 1, 3, 1, 1, 2, 2, 4],
    "price": [300, 1200, 100, 300, 1200, 100, 300, 100]
})

display(sales)
```

```python
#df["new_column"] = calculation
sales["total_sales"] = sales["quantity"] * sales["price"]

sales.head()
```

```python
sales.groupby("product")["quantity"].sum()
```

```python
sales.groupby("product")["quantity"].sum().reset_index()
```

```python
sales.groupby("customer")["total_sales"].sum().reset_index()
```

```python
display(sales)
```

```python
sales.groupby("state").agg(
    total_sales=(""total_sales","sum""),
    average_sales=("total_sales","mean"),
    total_quantity=("quantity", "sum")
).reset_index()
```

```python
sales.groupby("state").agg(
    total_sales=("total_sales","sum"),
    average_sales=("total_sales","mean"),
    total_quantity=("quantity", "sum"),
    order_count=("order_id","count")
).reset_index()
```

```python
sales.groupby(["state","product"]).agg(
    total_sales=("total_sales", "sum"),
    total_quantity=("quantity","sum")
).reset_index()
```

```python
product_summary = sales.groupby("product").agg(
    total_sales=("total_sales", "sum"),
    total_quantity=("quantity", "sum")
).reset_index()

display(product_summary)
```

```python
product_summary[product_summary["total_quantity"] >= 5]
```

Which customers have generated at least $1,000 in sales, and what are their total sales, total quantity purchased, and number of orders?

```python
display(sales)
```

```python
customer_sales = sales.groupby("customer").agg(
    total_sales = ("total_sales","sum"),
    total_quantity = ("quantity", "sum"),
    order_count = ("order_id", "count")
).reset_index()

display(customer_sales)
```

```python
customer_sales[customer_sales["total_sales"] >= 1000].sort_values("total_sales", ascending=False)
```

Hour 3: Combining DataFrames with merge()

```python
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "customer_name": ["Alex", "Maria", "James", "Sarah"],
    "state": ["CA", "CA", "TX", "NY"]
})

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005],
    "customer_id": [1, 1, 2, 3, 3],
    "product_id": [10, 11, 10, 12, 11],
    "quantity": [2, 1, 1, 3, 2]
})

products = pd.DataFrame({
    "product_id": [10, 11, 12],
    "product_name": ["Laptop", "Monitor", "Keyboard"],
    "price": [1200, 300, 100]
})

display(customers)
display(orders)
display(products)
```

Merge orders with customers

```python
# pd.merge(
#     left_dataframe,
#     right_dataframe,
#     on="key_column"
# )
```

```python
orders_customers = pd.merge(
    orders,
    customers,
    on="customer_id"
)

display(orders_customers)
```

```python
all_customers = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left")

display(all_customers)
```

```python
full_sales = pd.merge(
    orders_customers,
    products,
    on="product_id")

display(full_sales)
```

```python
full_sales["total_sales"] = full_sales["quantity"]*full_sales["price"]

full_sales[["order_id","customer_name","product_name","quantity","price","total_sales"]]
```

```python
orders_customers_2 = pd.merge(
    orders,
    customers,
    on="customer_id",
    validate="many_to_one"
)

display(orders_customers_2)

```

```python
orders_customers_2.shape
```

```python
sales_analysis = pd.merge(
    orders_customers_2,
    products,
    on="product_id",
    validate="many_to_one"
)

display(sales_analysis)
```

```python
sales_analysis.shape
```

```python
sales_analysis["total_sales"] = sales_analysis["quantity"] * sales_analysis["price"]
```

```python
sales_analysis.sort_values("total_sales",ascending=False)
```

```python
sales_analysis = sales_analysis.sort_values("total_sales",ascending=False)
display(sales_analysis)
```
