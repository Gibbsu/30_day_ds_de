Hour 2 — NumPy Statistics

```python
total_revenue = np.sum(total_sales)
average_sale = np.mean(total_sales)
median_sale = np.median(total_sales)
minimum_sale = np.min(total_sales)
maximum_sale = np.max(total_sales)
```

```python
print(total_revenue)
print(average_sale)
print(median_sale)
print(minimum_sale)
print(maximum_sale)
```

Standard Deviation

```python
sales_std = np.std(total_sales)
print(sales_std)
```

Variance

```python
sales_variance = np.var(total_sales)
print(sales_variance)
```

```python
sales_std ** 2
```

Percentiles

```python
q1 = np.percentile(total_sales, 25)
median = np.percentile(total_sales, 50)
q3 = np.percentile(total_sales, 75)
```

```python
print(q1)
print(median)
print(q3)
```

Interquartile Range (IQR).

```python
iqr = q3 - q1
```

```python
print(iqr)
```

Outlier Boundaries

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
```

```python
print(lower_bound)
print(upper_bound)
```

Find Outliers

```python
outliers = total_sales[
    (total_sales < lower_bound) |
    (total_sales > upper_bound)
]
print(outliers)
```

Hour 3 — Exploratory Data Analysis with Pandas

```python
sales_df = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Monitor", "Keyboard", "Laptop",
        "Monitor", "Keyboard", "Laptop", "Monitor",
        "Keyboard", "Laptop", "Monitor", "Keyboard"
    ],
    "quantity": [1, 2, 3, 1, 2, 5, 1, 3, 2, 1, 2, 4],
    "total_sales": [
        1200, 600, 300, 1200,
        600, 500, 1200, 900,
        200, 1200, 600, 400
    ]
})

display(sales_df)
```

describe()

```python
sales_df.describe()
```

Product-Level EDA

How do sales differ by product?

```python
product_summary = sales_df.groupby("product").agg(
    order_count=("order_id", "count"),
    average_sales=("total_sales", "mean"),
    median_sales=("total_sales", "median"),
    minimum_sales=("total_sales", "min"),
    maximum_sales=("total_sales", "max")
).reset_index()
```

```python
print(product_summary)
```

Calculate Range

```python
product_summary["sales_range"] = (
    product_summary["maximum_sales"] - product_summary["minimum_sales"]
)
```

Standard Deviation by Product

```python
product_variability = sales_df.groupby("product").agg(
    average_sales=("total_sales", "mean"),
    sales_std=("total_sales", "std")
).reset_index()

display(product_variability)
```

Find Unusually Large Sales

```python
q1 = sales_df["total_sales"].quantile(0.25)
q3 = sales_df["total_sales"].quantile(0.75)

iqr = q3 - q1
```

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
```

```python
potential_outliers = sales_df[
    (sales_df["total_sales"] < lower_bound) |
    (sales_df["total_sales"] > upper_bound)
]
```

Hour 4 — Apply EDA to Our Project

```python
project_sales = pd.read_csv("../data/sales_analysis_clean.csv")

display(project_sales)
```

How is sales performance different across products?

```python
project_product_summary = project_sales.groupby("product_name").agg(
    order_count=("order_id", "count"),
    total_revenue=("total_sales", "sum"),
    average_order_sales=("total_sales", "mean"),
    total_quantity=("quantity", "sum")
).reset_index()
```

Rank Products

Which products generate the most total revenue?

```python
project_product_summary = project_product_summary.sort_values(
    "total_revenue",
    ascending=False
)

display(project_product_summary)
```

Add Revenue Share

What percentage of total company revenue comes from each product?

```python
total_revenue = project_product_summary["total_revenue"].sum()
```

```python
print(total_revenue)
```

```python
project_product_summary["revenue_share"] = (
    project_product_summary["total_revenue"] / total_revenue
)
```

Outlier Check

Calculate Q1 and Q3

```python
q1 = project_sales["total_sales"].quantile(0.25)
q3 = project_sales["total_sales"].quantile(0.75)
```

calculate the IQR and the lower/upper outlier boundaries

```python
iqr = q3 - q1
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
```

use those boundaries to find the potential outliers

```python
project_outliers = project_sales[
    (project_sales["total_sales"] < lower_bound) |
    (project_sales["total_sales"] > upper_bound)
]

display(project_outliers)
```
